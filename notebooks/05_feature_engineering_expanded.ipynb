```python
import pandas as pd
import sys
sys.path.insert(0, '..')
from src.features.extractor import extract_features, get_feature_names

print("Imports successful")
print(f"Number of features now: {len(get_feature_names())}")
print(f"Features: {get_feature_names()}")
```

```python
df = pd.read_csv('../data/processed/combined_dataset.csv')
print(f"Loaded {len(df)} rows")
print(f"Label distribution:")
print(df['label'].value_counts())
```

```python
# This cell will take a long time — possibly 45-90 minutes for a large dataset
# Progress is shown every 10,000 URLs

print("Extracting features from all URLs...")
print("This will take a long time. Do not close VS Code.")

feature_rows = []
errors = 0

for i, row in df.iterrows():
    if i % 10000 == 0:
        print(f"  Processed {i}/{len(df)} URLs ({i/len(df)*100:.1f}%)")
    try:
        features = extract_features(str(row['url']))
        feature_rows.append(features)
    except Exception:
        feature_rows.append({name: 0 for name in get_feature_names()})
        errors += 1

print(f"\nDone! Errors: {errors}")
```

```python
features_df = pd.DataFrame(feature_rows)
features_df['label'] = df['label'].values

print(f"Shape: {features_df.shape}")
print(f"Label distribution:")
print(features_df['label'].value_counts())
print(f"\nSample of is_in_top_1million feature:")
print(features_df['is_in_top_1million'].value_counts())
```

```python
output_path = '../data/processed/features_dataset_expanded.csv'
features_df.to_csv(output_path, index=False)
print(f"Expanded features dataset saved to: {output_path}")
print(f"Total rows: {len(features_df)}")
```
