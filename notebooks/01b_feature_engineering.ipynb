```python
import pandas as pd
import sys
sys.path.insert(0, '..')
from src.features.extractor import extract_features, get_feature_names

print("Imports successful")
print(f"Number of features: {len(get_feature_names())}")
```

```python
df = pd.read_csv('../data/processed/cleaned_dataset.csv')
print(f"Loaded {len(df)} rows")
print(f"Columns: {df.columns.tolist()}")
print(df.head(3))
```

```python
print("Extracting features from all URLs...")
print("This may take 10-30 minutes. Progress shown every 10,000 URLs.")

feature_rows = []
errors = 0

for i, row in df.iterrows():
    if i % 10000 == 0:
        print(f"  Processed {i}/{len(df)} URLs ({i/len(df)*100:.1f}%)")
    try:
        features = extract_features(str(row['URL']))
        feature_rows.append(features)
    except Exception:
        feature_rows.append({name: 0 for name in get_feature_names()})
        errors += 1

print(f"\nDone! Errors: {errors}")
```

```python
features_df = pd.DataFrame(feature_rows)

# In PhiUSIIL dataset: 1 = legitimate, 0 = phishing
# Our code expects:     1 = phishing,   0 = legitimate
# So we flip the labels here
features_df['label'] = df['label'].apply(lambda x: 0 if x == 1 else 1).values

print(f"Shape: {features_df.shape}")
print()
print("Label distribution after flipping (1=phishing, 0=legitimate):")
print(features_df['label'].value_counts())
print()
print("First few rows:")
print(features_df.head())
```

```python
output_path = '../data/processed/features_dataset.csv'
features_df.to_csv(output_path, index=False)
print(f"Saved to: {output_path}")
```
