```python
import pandas as pd
import numpy as np
import os

print("Libraries loaded successfully")
print(f"Pandas version: {pd.__version__}")
```

```python
df = pd.read_csv('../data/raw/PhiUSIIL_Phishing_URL_Dataset.csv')

print(f"Dataset shape: {df.shape}")
print(f"\nColumn names: {df.columns.tolist()}")
print(f"\nFirst 5 rows:")
print(df.head())
```

```python
print("Label distribution:")
print(df['label'].value_counts())
```

```python
print("Missing values per column:")
print(df.isnull().sum())
```

```python
initial_count = len(df)
df = df.dropna()
final_count = len(df)

print(f"Rows before cleaning: {initial_count}")
print(f"Rows after removing missing values: {final_count}")
print(f"Removed {initial_count - final_count} rows")
```

```python
before_dedup = len(df)
df = df.drop_duplicates()
after_dedup = len(df)

print(f"Rows before removing duplicates: {before_dedup}")
print(f"Rows after removing duplicates: {after_dedup}")
print(f"Removed {before_dedup - after_dedup} duplicate rows")
```

```python
print("Sample URLs from the dataset:")
for url in df['url'].head(10).tolist():
    print(f"  {url}")
```

```python
output_path = '../data/processed/cleaned_dataset.csv'
df.to_csv(output_path, index=False)

print(f"Cleaned dataset saved to: {output_path}")
print(f"Final size: {df.shape[0]} rows, {df.shape[1]} columns")
```
